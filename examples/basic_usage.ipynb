````python
import numpy as np
from rtgym import RatatouGym
````

## Creating Arena
### Basic Arena

````python
gym = RatatouGym(
   temporal_resolution=50,
   spatial_resolution=1,
)
gym.init_arena_map(shape='rectangle', dimensions=[100, 100])
_ = gym.arena.vis()
````

### Define Custom Arena

````python
gym = RatatouGym(
   temporal_resolution=50,
   spatial_resolution=1,
)
arena_map = np.ones((300, 150))
arena_map[10:290, 10:140] = 0
arena_map[160:170, 50:150] = 1
gym.arena.set_arena_map(arena_map)
_ = gym.arena.vis()
````

## Initializing Agent's Behavior
#### Agent's Behavior Can be Configured Using a Dictionary
Template for the behavior configuration dictionary:

```python
{
    'avoid_boundary_dist': ,
    'max_velocity': 20,
    'random_drift_magnitude': 0.03,
    'switch_direction_prob': 0.03,
    'switch_velocity_prob': 0.1,
    'velocity_mean': 20,
    'velocity_sd': 10,
    'pause_prob': 0.2
}
```

````python
behavior_config = {
    'avoid_boundary_dist': 80,
    'max_velocity': 20,
    'random_drift_magnitude': 0.03,
    'switch_direction_prob': 0.03,
    'switch_velocity_prob': 0.1,
    'velocity_mean': 20,
    'velocity_sd': 10,
    'pause_prob': 0.2
}
gym.agent.behavior.init_from_profile(behavior_config)
traj = gym.agent.random_traverse(duration=20, batch_size=1)
````

````python
gym.vis_traj(traj)
````

````python
neuron_profiles = {
    "grid": {
        "type": "grid_cell",
        "n_cells": 100
    }
}
gym.agent.init_neurons(neuron_profiles)
````
